### Array Reshaping

```python
arr = np.arange(12)
print("Original array : ", arr)

reshaped = arr.reshape((3,4))
print("Reshaped array : \n", reshaped)

flattened = reshaped.flatten()
print("flattened array : \n", flattened)

#ravel (returns view instead of copy. [it returns original])
raveled = reshaped.ravel()
print("\n raveled array : ",raveled)

#Transpose
transpose = reshaped.T
print("\n Transpose array : \n", transpose)

```

### Filtering Array

```python
numbers = np.array([1,2,3,4,5,49,34,56,75,34])

#numpy allows you to use expressions.

even_numbers = numbers [numbers % 2 == 0]
print("even numbers : ", even_numbers)
```

### Filter with Mask

```python
#mask represents to an expression for evaluation
mask = numbers > 5
print("Numbers greater than 5 : ", numbers[mask])#mask is an expression [numbers > 5]
```

### Fancy indexing vs np.where()

```python
#accessing with different indices
indices = [0,2,4]
print("accessed array values using indexs-list : indices -->",indices,"values ==> ",numbers[indices])

##access array values by checking conditions using where keyword
where_result = np.where(numbers>5) ##checking the condition if numbers array contains indexes more than 5

#lets see the result of where condtion
print("where condition output : ", where_result) #output with array index values.

#now we can use this where_result to get the actual values of an array
#where is also similar to mask with slight difference
print("Acessing array values using where : ",numbers[where_result])

```

```python
condition_array = np.where(numbers>5) #this where method applies the condition to numbers array. returns array indexes.

print("condition_array : ", condition_array)
print("we can see the array retruned its index values.")
```

```python
#understanding where method deeply.
#syntax : np.where(condition, true, false)
#it is very simple if the condtion true will do operation with the true case
#otherwise will perform the operation with false case

example_array = np.where(numbers>5 , numbers*5, numbers)
print(example_array      )
```

### Adding and Removing data

```python
arr1 = np.array([1,2,3])
arr2 = np.array([4,5,6])

combined_arr = arr1 + arr2
print("combined array : ", combined_arr)
print("this is not the right way to combine two arrays,the above operation is adding the elements of arras.\n")

print("CORRECT METHOD")
combined_arr = np.concatenate((arr1,arr2))
print("combined array : ", combined_arr)
```

### Array Compatibility

```python
original_arr = np.array([[1,2],[3,4]])
new_row = np.array([[5,6]])

#vstack mehod adds row , it accepts arugments as tuples.
with_new_row = np.vstack((original_arr,new_row))
print("row added horizontally : \n" , with_new_row)

new_col = np.array([[7],[8]])

#hstack method adds columns.
with_new_col = np.hstack((original_arr,new_col))
print("\n column added  vertically: \n", with_new_col)
```

### Phase-3 Advanced Operations with Business examples

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Data structure: [restaurant_id, 2021, 2022, 2023, 2024]
sales_data = np.array([
    [1, 150000, 180000, 220000, 250000],   # Paradise Biryani
    [2, 120000, 140000, 160000, 190000],   # Beijing Bites
    [3, 200000, 230000, 260000, 300000],   # Pizza Hub
    [4, 180000, 210000, 240000, 270000],   # Burger Point
    [5, 160000, 185000, 205000, 230000]    # Chai Point
])

print("----Zomato Sales Analysis------")
print("\n Sales data shape", sales_data.shape)
print("\n Sampled data for 1st 3 restaurants : \n", sales_data[0:3])
print("\n Sampled data for 1st 3 restaurants with Fancy access : \n", sales_data[:3])
```

```python

print("\n Acces Rows & columns : \n", sales_data[:,1:])
```

 Total Sales per Year

```python


print("Total Sales : ", np.sum(sales_data, axis=0))

yearly_total = np.sum(sales_data[:,1:], axis=0)
print("Yearly Sales Coloumn wise : \n", yearly_total) # This will calculate in Column wise 
#150000 
#120000 
#200000
#180000 
#160000

yearly_total = np.sum(sales_data[:,1:], axis=1) # This will calculate in row wise 150000+180000+220000+250000
print("Yearly Sales Row wise : \n", yearly_total)
```

```python
    # [1, 150000, 180000, 220000, 250000],   # Paradise Biryani
    # [2, 120000, 140000, 160000, 190000],   # Beijing Bites
    # [3, 200000, 230000, 260000, 300000],   # Pizza Hub
    # [4, 180000, 210000, 240000, 270000],   # Burger Point
    # [5, 160000, 185000, 205000, 230000]    # Chai Point


#Minimum Sales
#in min method, axis values works opposite to the normal methods
#0=> Rows, 1 => COl
min_sales = np.min(sales_data[:,1:],axis=0) #this method returns the minimum value in the columnwise

print("Minimum sales in the year in col wise : ",min_sales)

min_sales = np.min(sales_data[:,1:],axis=1) #this method returns the minimum value in the Row wise
print("Minimum sales in the year in row wise : ",min_sales)

```

```python
#Maximum Sales Per Year
max_sales = np.max(sales_data[:,1:],axis=0)
print("Max Sales per year column wise : ", max_sales)
```

```python
#Average Sales per restaurant
avg_sales = np.mean(sales_data[:,1:],axis=1)
print("average sales per restaurant : ", avg_sales)
```

```python
#Cumulative Sum
cum_sum_sales = np.cumulative_sum(sales_data[:,1:],axis=1)

print("Cumulative Sales : \n", cum_sum_sales)

plt.figure(figsize=(10,6))
plt.plot(np.mean(cum_sum_sales, axis=0))
plt.title("Average Cumulative Sales across all restaurants ")
plt.xlabel("Years")
plt.ylabel("Sales")
plt.grid(True)
plt.show()
```
