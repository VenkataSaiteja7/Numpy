### Array Reshaping

```python
arr = np.arange(12)
print("Original array : ", arr)

reshaped = arr.reshape((3,4))
print("Reshaped array : \n", reshaped)

flattened = reshaped.flatten()
print("flattened array : \n", flattened)

#ravel (returns view instead of copy. [it returns original])
raveled = reshaped.ravel()
print("\n raveled array : ",raveled)

#Transpose
transpose = reshaped.T
print("\n Transpose array : \n", transpose)

```

### Sorting

```python
unsortedArr = np.array([2,4,2,77,44,78,23,45,67,23,12,34,64]
                       )
print("Sorted Array : ", np.sort(unsortedArr))

#sorting 2 dimensional array
arr_2d_unsorted = np.array([[3,4],
                            [90,56],
                            [34,6]])
print("Sort 2D array by column : ", np.sort(arr_2d_unsorted,axis=0))
print("Sort 2D array by row : ", np.sort(arr_2d_unsorted,axis=1))
```

### Filtering Array

```python
numbers = np.array([1,2,3,4,5,49,34,56,75,34])

#numpy allows you to use expressions.

even_numbers = numbers [numbers % 2 == 0]
print("even numbers : ", even_numbers)
```

### Filter with Mask

```python
#mask represents to an expression for evaluation
mask = numbers > 5
print("Numbers greater than 5 : ", numbers[mask])#mask is an expression [numbers > 5]
```

### Fancy indexing vs np.where()

```python
#accessing with different indices
indices = [0,2,4]
print("accessed array values using indexs-list : indices -->",indices,"values ==> ",numbers[indices])

##access array values by checking conditions using where keyword
where_result = np.where(numbers>5) ##checking the condition if numbers array contains indexes more than 5

#lets see the result of where condtion
print("where condition output : ", where_result) #output with array index values.

#now we can use this where_result to get the actual values of an array
#where is also similar to mask with slight difference
print("Acessing array values using where : ",numbers[where_result])

```

```python
condition_array = np.where(numbers>5) #this where method applies the condition to numbers array. returns array indexes.

print("condition_array : ", condition_array)
print("we can see the array retruned its index values.")
```

```python
#understanding where method deeply.
#syntax : np.where(condition, true, false)
#it is very simple if the condtion true will do operation with the true case
#otherwise will perform the operation with false case

example_array = np.where(numbers>5 , numbers*5, numbers)
print(example_array      )
```

### Adding and Removing data

```python
arr1 = np.array([1,2,3])
arr2 = np.array([4,5,6])

combined_arr = arr1 + arr2
print("combined array : ", combined_arr)
print("this is not the right way to combine two arrays,the above operation is adding the elements of arras.\n")

print("CORRECT METHOD")
combined_arr = np.concatenate((arr1,arr2))
print("combined array : ", combined_arr)
```

```python
original_arr = np.array([[1,2],[3,4]])
new_row = np.array([[5,6]])

#vstack mehod adds row , it accepts arugments as tuples.
with_new_row = np.vstack((original_arr,new_row))
print("row added horizontally : \n" , with_new_row)

new_col = np.array([[7],[8]])

#hstack method adds columns.
with_new_col = np.hstack((original_arr,new_col))
print("\n column added  vertically: \n", with_new_col)
```
